## The Impact of Artificial Intelligence Tools on Students' Learning and Academic Performance

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
sns.set()
```

```python
df = pd.read_csv('Clean_Dataset_AI_Students.csv')
df.head()
```

## Data Cleaning

```python
df.columns = df.columns.str.strip()
df = df.drop_duplicates()
df.isnull().sum()
```

## Analysis 1: AI Usage Distribution

```python
df['Usage_Frequency'].value_counts().plot(kind='bar', title='AI Usage Distribution')
plt.xlabel('Usage')
plt.ylabel('Number of Students')
plt.show()
```

## Analysis 2: Did Students Improve Using AI?

```python
df['Grades_Improved'].value_counts().plot(kind='pie', autopct='%1.1f%%', title='Did Students Improve Using AI?')
plt.ylabel('')
plt.show()
```

## Analysis 3: Grades Distribution

```python
plt.hist(df['Grade'], bins=10)
plt.title('Grades Distribution')
plt.xlabel('Grades')
plt.ylabel('Frequency')
plt.show()
```
